# 数据集民族筛选

对 `基础信息.tsv` 进行人群名称校正，然后将目标民族筛选出来，供后续骨架树筛选使用。

```python
import pandas as pd
```

## 一、加载原始基础信息，生成 df_校正后

```python
df_基础信息 = pd.read_csv("../meta/基础信息.tsv", sep="\t", low_memory=False)

# 标准化人群名称（校正拼写差异，统一蒙古族、维吾尔族等别名）
population_rename_map = {
    "Mogolian": "Mongolian",
    "Uighur":   "Uygur",
    "Shui":     "Sui",
    "Ewenki":   "Evenki",
    "Xibo":     "Xibe",
    "Chosen":   "Korean",
    "Russ":     "Russian",
    "Kazakh":   "Kazak",
    "Kirgiz":   "Kyrgyz",
    "Yugu":     "Yugur",
    "Mulam":    "Mulao",
    "Baoan":    "Bonan",
    "Man":      "Manchu",
    "Mongol":   "Mongolian",
    "Zang":     "Tibetan",
    "Monpa":    "Menba",
    "Chinese":  "Han",
}

df_校正后 = df_基础信息.copy()
df_校正后["Population"] = df_校正后["Population"].replace(population_rename_map)
df_校正后
```
